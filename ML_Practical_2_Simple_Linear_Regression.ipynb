# Machine Learning – Practical 2: Simple Linear Regression (Temperature vs Year)

**Aim:** To predict annual temperature from the year using Simple Linear Regression.

**Objective:**
- Explore the temperature dataset
- Visualise the relationship between year and temperature
- Train a Linear Regression model
- Evaluate it using MSE, MAE, RMSE and R²

**Theory:** Simple Linear Regression fits a straight line y = mx + c between one input (x) and one output (y) by minimising squared errors. Here x = YEAR and y = ANNUAL temperature.

## 1. Import Libraries & Load Data

**Aim:** Load the temperature dataset.

**Objective:** Import libraries and read the CSV.

**Theory:** Pandas loads the data, Seaborn/Matplotlib visualise it and scikit-learn provides the regression model.

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

data = pd.read_csv(
    r"C:\Users\staff\Downloads\archive(1)\temperatures.csv"
)
df = data

print(data.head())
```

## 2. Data Exploration

**Aim:** Understand the dataset.

**Objective:** View statistics, last rows, shape and data types.

**Theory:** `describe()` gives summary statistics, `shape` gives rows × columns and `info()` shows data types and non-null counts.

```python
data.describe()
```

```python
data.tail()
```

```python
print(type(data))
print(data.shape)
data.info()
```

## 3. Check for Missing / Zero Values

**Aim:** Ensure data quality.

**Objective:** Count zero values and null values in every column.

**Theory:** Missing or zero values can distort a regression model, so they must be checked before training.

```python
column = data
count = column[column == 0].count()
print(count)
```

```python
data.isnull().sum()
```

## 4. Define Input/Output & Visualise

**Aim:** Choose variables and view their relationship.

**Objective:** Set x = YEAR, y = ANNUAL and draw scatter plots.

**Theory:** A scatter plot shows whether a linear relationship exists. An upward trend suggests temperature increases over time.

```python
# Input variable
x = data["YEAR"]

# Output variable
y = data["ANNUAL"]

plt.plot(x, y, "o")
plt.xlabel("Year")
plt.ylabel("Annual Temperature")
plt.title("Temperature vs Year")
plt.show()
```

```python
sns.scatterplot(x=x, y=y, data=df)
plt.title("Temperature vs Year")
plt.xlabel("Year")
plt.ylabel("Annual Temperature")
plt.show()
```

## 5. Reshape Data & Train-Test Split

**Aim:** Prepare data for scikit-learn.

**Objective:** Convert x to a 2D array and split into 75% train / 25% test.

**Theory:** scikit-learn expects features as a 2D array (samples × features). The split lets us test the model on unseen data.

```python
x = x.values
x = x.reshape(117, 1)
print(x.shape, type(x))

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.25
)

print(f"x Training dataset: {x_train.shape}")
print(f"y Training dataset: {y_train.shape}")
print(f"x test dataset: {x_test.shape}")
print(f"y test dataset: {y_test.shape}")
```

## 6. Train the Model

**Aim:** Fit the regression line.

**Objective:** Train Linear Regression and read the slope and intercept.

**Theory:** The model learns the slope (coefficient) and intercept: temperature = slope × year + intercept. The slope shows the yearly temperature change.

```python
model = LinearRegression()
model.fit(x_train, y_train)

print("Coefficient:", model.coef_)
print("Intercept:", model.intercept_)
```

## 7. Prediction & Visualisation

**Aim:** Predict on test data and plot the fit.

**Objective:** Predict temperatures for the test years and draw the regression line.

**Theory:** The fitted line summarises the overall warming trend; points far from it are residual errors.

```python
y_pred = model.predict(x_test)
print(y_pred.shape)

plt.scatter(x_train, y_train, color="blue")
plt.plot(x_test, y_pred, color="red", linewidth=3)

plt.title("Temperature vs Year")
plt.xlabel("Year")
plt.ylabel("Temperature")
plt.show()
```

```python
sns.regplot(
    x=x_train.ravel(),
    y=y_train
)

plt.title("Linear Regression: Temperature vs Year")
plt.xlabel("Year")
plt.ylabel("Annual Temperature")
plt.show()
```

## 8. Model Evaluation

**Aim:** Measure prediction quality.

**Objective:** Compute MSE, MAE, RMSE and R².

**Theory:** MSE/RMSE penalise large errors, MAE is the average absolute error, and R² is the share of variance explained (closer to 1 is better).

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print(f"MSE: {mean_squared_error(y_test, y_pred)}")
print(f"MAE: {mean_absolute_error(y_test, y_pred)}")
print(f"RMSE: {np.sqrt(mean_squared_error(y_test, y_pred))}")
print(f"R-Square: {r2_score(y_test, y_pred)}")
```

## Conclusion

Annual temperature shows a clear upward trend with year (slope ≈ 0.014 °C per year). The low R² (≈ 0.22 in the original run) shows year alone explains only part of the variation, and the split is random, so exact values change between runs.
