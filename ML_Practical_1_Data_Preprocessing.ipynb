# Machine Learning – Practical 1: Data Preprocessing & Feature Engineering (Loan Dataset)

**Aim:** To clean and prepare the loan dataset so it is ready for machine learning.

**Objective:**
- Explore the dataset and find missing values
- Impute missing values
- Encode categorical variables
- Scale numerical features
- Save the cleaned dataset

**Theory:** Raw data usually has missing values, text categories and features on different scales. Preprocessing (imputation, encoding, scaling) converts it into a clean numeric form so models can learn properly.

## 1. Import Libraries & Load Data

**Aim:** Load the loan dataset and inspect it.

**Objective:** Read the CSV and view its first rows.

**Theory:** Pandas loads tabular data into a DataFrame; `head()` gives a quick look at its structure.

```python
import pandas as pd
import numpy as np

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder, StandardScaler

data = pd.read_csv(r"C:\Users\staff\Downloads\loanData (1).csv")

print("First Five Records")
print(data.head())
```

## 2. Exploratory Data Analysis

**Aim:** Understand the dataset's structure and quality.

**Objective:** Check data types, shape, summary statistics and missing values.

**Theory:** `info()`, `describe()` and `isnull().sum()` reveal data types, value ranges (mean, std, min, max) and which columns need cleaning.

```python
print("\nDataset Information")
print(data.info())

print("\nDataset Shape")
print(data.shape)

print("\nSummary Statistics")
print(data.describe())

print("\nMissing Values")
print(data.isnull().sum())
```

## 3. Handling Missing Values

**Aim:** Fill in missing data.

**Objective:** Impute numerical columns with the median and categorical columns with the most frequent value.

**Theory:** Median imputation is robust to outliers for numeric data; mode (most frequent) imputation suits categorical data. Dropping rows would waste data.

```python
num_cols = ['LoanAmount', 'Loan_Amount_Term', 'Credit_History']

num_imputer = SimpleImputer(strategy='median')
data[num_cols] = num_imputer.fit_transform(data[num_cols])

# Categorical Columns
cat_cols = [
    'Gender',
    'Married',
    'Dependents',
    'Self_Employed'
]

cat_imputer = SimpleImputer(strategy='most_frequent')
data[cat_cols] = cat_imputer.fit_transform(data[cat_cols])

print("\nMissing Values After Imputation")
print(data.isnull().sum())
```

## 4. Drop ID Column & Encode Categorical Variables

**Aim:** Convert text categories into numbers.

**Objective:** Remove `Loan_ID` and apply Label Encoding to categorical columns.

**Theory:** `Loan_ID` is a unique identifier with no predictive value. Label Encoding maps each category to an integer (e.g. Male/Female → 1/0) since ML models need numeric input.

```python
if 'Loan_ID' in data.columns:
    data.drop('Loan_ID', axis=1, inplace=True)

encoder = LabelEncoder()

categorical_columns = [
    'Gender',
    'Married',
    'Dependents',
    'Education',
    'Self_Employed',
    'Property_Area',
    'Loan_Status'
]

for column in categorical_columns:
    if column in data.columns:
        data[column] = encoder.fit_transform(
            data[column].astype(str)
        )

print("\nDataset After Encoding")
print(data.head())
```

## 5. Feature Scaling

**Aim:** Bring numerical features to a common scale.

**Objective:** Standardise numeric columns to mean 0 and standard deviation 1.

**Theory:** StandardScaler computes z = (x − mean) / std. It prevents large-valued features (e.g. income) from dominating smaller ones.

```python
scaler = StandardScaler()

numerical_columns = [
    'ApplicantIncome',
    'CoapplicantIncome',
    'LoanAmount',
    'Loan_Amount_Term',
    'Credit_History'
]

numerical_columns = [
    col for col in numerical_columns if col in data.columns
]

data[numerical_columns] = scaler.fit_transform(
    data[numerical_columns]
)

print("\nDataset After Scaling")
print(data.head())
```

## 6. Final Check & Save

**Aim:** Verify and export the cleaned data.

**Objective:** Confirm no missing values remain and save to CSV.

**Theory:** A final check ensures the dataset is fully numeric and complete before it is used for modelling.

```python
print("\nFinal Dataset Information")
data.info()

print("\nFirst Five Rows of Processed Dataset")
print(data.head())

print("\nRemaining Missing Values")
print(data.isnull().sum())

output_path = r"C:\Users\staff\Downloads\Loan_Prediction_Cleaned.csv"

data.to_csv(output_path, index=False)

print("\n======================================")
print("Feature Engineering Completed")
print("Clean Dataset Saved Successfully")
print("Saved at:", output_path)
print("======================================")
```

## Conclusion

The 614 × 13 loan dataset was cleaned: all missing values were imputed, `Loan_ID` was dropped, categories were encoded and numeric features were standardised, giving a clean 614 × 12 dataset.
